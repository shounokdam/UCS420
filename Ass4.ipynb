```python
import pandas as pd

# ---------------------------------------------------------
# Q1:

# ---------------------------------------------------------

# Enter your college roll number here
roll_number = input("Enter your roll number: ")

# Convert roll number into string
roll_number = str(roll_number)

# Take last two digits
last_two_digits = roll_number[-2:]

# Fixed FAQ entries
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Categories
categories = ["billing", "account", "general"]

# Personalized entries
personalized_entries = []

for digit in last_two_digits:
    d = int(digit)
    category = categories[d % 3]

    if category == "billing":
        question = "how can i check my payment status"
        answer = "You can check your payment status from the Billing section."
        keywords = "payment status billing transaction"

    elif category == "account":
        question = "how do i update my registered mobile number"
        answer = "Go to Account Settings and update your registered mobile number."
        keywords = "mobile number update account"

    else:
        question = "where can i find general help"
        answer = "You can find general help in the Help and Support section."
        keywords = "help support information"

    personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

# Combine fixed + personalized entries
all_entries = fixed_entries + personalized_entries

# Create DataFrame
df = pd.DataFrame(all_entries)

print("\nQ1: Final 6-row DataFrame")
print(df)


# ---------------------------------------------------------
# Q2: Generate and Score a Hypothesis
# ---------------------------------------------------------

def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        # Combine question and keywords
        text = row["question"].lower() + " " + row["keywords"].lower()

        text_words = set(text.split())

        # Count matching words
        matching_words = query_words.intersection(text_words)

        score = len(matching_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    # Rank by confidence score
    results = sorted(results, key=lambda x: x["score"], reverse=True)

    return results


# Demonstrate Q2
query = input("\nQ2: Enter a query: ")

results = score_query(query, df)

print("\nMatching entries ranked by confidence:")

if results:
    for result in results:
        print(result)
else:
    print("No matching entries found.")


# ---------------------------------------------------------
# Q3: same_category()
# ---------------------------------------------------------

def same_category(category_name, df):
    return df[df["category"] == category_name]


# Use category of first personalized entry
personalized_category = personalized_entries[0]["category"]

print("\nQ3: Entries belonging to category:",
      personalized_category)

print(same_category(personalized_category, df))


# ---------------------------------------------------------
# Q4: Add a new keyword and save CSV
# ---------------------------------------------------------

print("\nQ4: Choose an FAQ entry")

# Pick first entry for demonstration
entry_index = 0

print("Selected question:")
print(df.loc[entry_index, "question"])

new_keyword = input("Enter a new keyword to add: ")

# Add new keyword
df.loc[entry_index, "keywords"] = (
    df.loc[entry_index, "keywords"] + " " + new_keyword
)

# Save CSV
filename = roll_number + "_faq_data.csv"

df.to_csv(filename, index=False)

print("\nUpdated DataFrame:")
print(df)

print("\nCSV saved as:", filename)


# ---------------------------------------------------------
# Q5: Count FAQ entries per category using groupby
# ---------------------------------------------------------

print("\nQ5: Number of FAQ entries per category:")

category_counts = df.groupby("category").size()

print(category_counts)


# ---------------------------------------------------------
# Q6: Modified scoring function with tie handling
# ---------------------------------------------------------

def score_query_with_ties(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        text = row["question"].lower() + " " + row["keywords"].lower()

        text_words = set(text.split())

        matching_words = query_words.intersection(text_words)

        score = len(matching_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    # No matches
    if not results:
        print("No matching entries found.")
        return []

    # Sort by score
    results.sort(key=lambda x: x["score"], reverse=True)

    # Highest score
    highest_score = results[0]["score"]

    # Get all entries having highest score
    highest_matches = [
        result for result in results
        if result["score"] == highest_score
    ]

    # Tie handling
    if len(highest_matches) > 1:
        print("\nTIE FOUND!")
        print("Multiple entries have the highest confidence score:")

        for result in highest_matches:
            print(result)

    else:
        print("\nNo tie. Best matching entry:")
        print(highest_matches[0])

    return results


# ---------------------------------------------------------
# Q6 Demonstration 1: Query producing a tie
# ---------------------------------------------------------

print("\nQ6 Demonstration 1: Tie query")

tie_results = score_query_with_ties("fee", df)


# ---------------------------------------------------------
# Q6 Demonstration 2: Query without a tie
# ---------------------------------------------------------

print("\nQ6 Demonstration 2: Non-tie query")

non_tie_results = score_query_with_ties("password", df)
```
