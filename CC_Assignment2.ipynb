```python
#Ques3
import random
import string as s
random.seed(1024160092)

numbers = [random.randint(100,900) for _ in range(100)]
print(numbers)

odd_nums=[num for num in numbers if num%2!=0]
print("Odd numbers:",odd_nums)

even_nums=[num for num in numbers if num%2==0]
print("Even numbers:",even_nums)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime numbers:",prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

frequency = {}
for i in numbers:
    frequency[i] = frequency.get(i, 0) + 1
most_frequent = max(frequency, key=frequency.get)
most_frequent_count = frequency[most_frequent]
print("Most frequent number is: ", most_frequent, "with count: ", most_frequent_count)
```

```python
#Ques 4
roll_no="1024160092"
digits = [int(digit) for digit in roll_no[:8]]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:",A)
print("Set B:",B)

print(A.union(B))

print(A.intersection(B))

print("A.diffrence on B",A.difference(B))
print("B.diffrence on A",B.difference(A))
#Difference gives the elemnts that are present in one set and not the other but symmetric_difference() gives elements present in either set but not in both

print("A.symmetric_difference on B",A.symmetric_difference(B))

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

x=int(input("Enter a number"))
A.discard(x)
print(A)
#discard is safer than remove because if the specified value does not exist in the set then discard doesnt give an error whereas remove does
```
